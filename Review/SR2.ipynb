```
public interface IReportable
{
    void GenerateReport();
}
public interface ITeamLeader
{
    void GetTeamMembers();
}

public class Employee
{
    public string Name { get; set; }
    public string Position { get; set; }
    public double Salary { get; set; }

    public Employee(string name, string position, double salary)
    {
        Name = name;
        Position = position;
        Salary = salary;
    }

    public virtual void PrintInfo()
    {
        Console.WriteLine($"Сотрудник: {Name}\nДолжность: {Position}\nЗаработная плата: {Salary}руб.");
    }
}

public class Manager : Employee
{
    protected string Departament { get; set; }

    public Manager(string name, string position, double salary, string departament) 
    : base(name, position, salary)
    {
        Departament = departament;
    }

    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"Менеджер {Name} управляет отделом: {Departament}");
    }
}

public class ProjectManager : Manager, IReportable, ITeamLeader
{
    protected string ProjectName { get; set; }
    private List<string> TeamMembers = new List<string>();


    public ProjectManager(string name, string position, double salary,
     string departament, string projectname) : base(name, position, salary, departament)
    {
        ProjectName = projectname;
    }

    public void AddTeamMember(string NameMember)
    {
        TeamMembers.Add(NameMember);
    }
    
    public void GenerateReport()
    {
        Console.WriteLine($"ОТЧЕТ ПО ПРОЕКТУ\nНазвание проекта: {ProjectName}\nМенеджер проекта: {Name}, должность в компании - {Position}, заработная плата - {Salary}руб.\nКоличество сотрудников, работающих над проектом: {TeamMembers.Count}");
    }

    public void GetTeamMembers()
    {
        if(TeamMembers.Count != 0)
        {
            Console.WriteLine("Список команды:");
            foreach(string m in TeamMembers)
            {
                Console.Write(m + ", ");
            }
            Console.Write("\n");
        }
        else
        {
            Console.WriteLine("Над данным проектом не работает ни один из сотрудников");
        }
    }

    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"Проектный менеджер {Name} руководит разработкой проекта: {ProjectName}");
    }
}

Employee employee1 = new Employee("Алексей", "Тестировщик", 99.235);
employee1.PrintInfo();
Console.WriteLine(new string('=', 40));
Manager manager1 = new Manager("Виктор", "Backend-разработчик", 124.235, "Отдел разработки");
manager1.PrintInfo();
Console.WriteLine(new string('=', 40));
ProjectManager pm1 = new ProjectManager("Андрей", "Директор", 156.999, "Руководитель команды", "Проект Феникс");
pm1.PrintInfo();
Console.WriteLine(new string('=', 40));
//МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ
pm1.AddTeamMember("Иван");
pm1.AddTeamMember("Артём");
pm1.AddTeamMember("Николай");
pm1.AddTeamMember("Мария");
pm1.AddTeamMember("Григорий");
pm1.GetTeamMembers();
pm1.GenerateReport();
```
