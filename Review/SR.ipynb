```
public class Plant
{
    public string Name { get; set; }
    public string Type { get; set; }
    public float Height { get; set; }

    public Plant(string name, string type, float height)
    {
        Name = name;
        Type = type;
        Height = height;
    }

    public virtual void Bloom()
    {
        Console.WriteLine($"Растение '{Name}' цветёт.");
    }

    public void Grow(float growth)
    {
        float oldHeight = Height;
        Height += growth;
        Console.WriteLine($"'{Name}' вырос(ла) с {oldHeight} см до {Height} см (+{growth} см).");
    }

    public void ShowInfo()
    {
        Console.WriteLine($"[{Type}] {Name} | Высота: {Height} см");
    }
}

public class Flower : Plant
{
    public string Color { get; set; }
    public int PetalCount { get; set; }

    public Flower(string name, float height, string color, int petalCount)
        : base(name, "Цветок", height)
    {
        Color = color;
        PetalCount = petalCount;
    }

    public override void Bloom()
    {
        Console.WriteLine($"Цветок '{Name}' распускается и вкусно пахнет!");
    }

    public void Wilt()
    {
        Console.WriteLine($"Цветок '{Name}' начинает засыхать...");
    }

    public new void ShowInfo()
    {
        Console.WriteLine($"[Цветок] {Name} | Цвет: {Color}, Лепестков: {PetalCount}, Высота: {Height} см");
    }
}

public class Tree : Plant
{
    public int Age { get; set; }
    public string LeafType { get; set; }

    public Tree(string name, float height, int age, string leafType)
        : base(name, "Дерево", height)
    {
        Age = age;
        LeafType = leafType;
    }

    public override void Bloom()
    {
        Console.WriteLine($"Дерево '{Name}' покрывается листьями!");
    }

    public void GrowYears(int years)
    {
        Age += years;
        Console.WriteLine($"Дерево '{Name}' стало старше на {years} л. Новый возраст: {Age} лет.");
    }

    public new void ShowInfo()
    {
        Console.WriteLine($"[Дерево] {Name} | Возраст: {Age} лет, Листья: {LeafType}, Высота: {Height} см");
    }
}


Console.WriteLine("САД:");

Flower rose = new Flower("Роза", 45.0f, "Красный", 25);
Tree oak = new Tree("Дуб", 350.0f, 50, "Лиственные");

rose.ShowInfo();
oak.ShowInfo();
Console.WriteLine();

        
Plant[] garden = { rose, oak };
Console.WriteLine("Цветение (полиморфизм)");
foreach (Plant plant in garden)
{
    plant.Bloom();
}
Console.WriteLine();

Console.WriteLine("Рост");
rose.Grow(5.5f);
oak.Grow(20.0f);
Console.WriteLine();

Console.WriteLine("Специфичные методы");
rose.Wilt();
oak.GrowYears(10);
Console.WriteLine();

Console.WriteLine("Итоговое состояние");
rose.ShowInfo();
oak.ShowInfo();
```
