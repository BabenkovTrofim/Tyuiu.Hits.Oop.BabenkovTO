```
public class Product
{
    private string name;
    private decimal price;
    private int quantity;
    protected string category;
    internal string description;

    public string Name { get => name; set => name = value; }
    public decimal Price { get => price; set => price = value; }
    public int Quantity { get => quantity; set => quantity = value; }
    public string Category { get => category; set => category = value; }

    public Product(string name, decimal price, int quantity, string category, string description)
    {
        this.name = name;
        this.price = price;
        this.quantity = quantity;
        this.category = category;
        this.description = description;
    }

    public decimal GetTotalPrice() => price * quantity;

    public void UpdateQuantity(int amount) { quantity += amount; }

    public string GetProductInfo() =>
        $"{name} | {category} | {price}₽ x {quantity} = {GetTotalPrice()}₽ | {description}";
}

public class Seller
{
    private string name;
    private string employeeId;
    protected decimal salary;
    internal string contactInfo;
    private List<Product> products = new List<Product>();

    public string Name { get => name; set => name = value; }
    public string EmployeeId { get => employeeId; set => employeeId = value; }
    public decimal Salary { get => salary; set => salary = value; }
    public List<Product> Products => products;

    public Seller(string name, string employeeId, decimal salary, string contactInfo)
    {
        this.name = name;
        this.employeeId = employeeId;
        this.salary = salary;
        this.contactInfo = contactInfo;
    }

    public void AddProduct(Product product) { products.Add(product); }

    public void SellProduct(Product product, int qty)
    {
        if (product.Quantity >= qty) { product.UpdateQuantity(-qty); Console.WriteLine($"Продано {qty} шт. '{product.Name}'"); }
        else Console.WriteLine($"Недостаточно '{product.Name}' на складе!");
    }

    public string GetSellerInfo() => $"{name} (ID: {employeeId}) | Зарплата: {salary}₽ | {contactInfo}";
}

public class Store
{
    private string storeName;
    private string location;
    public string storeHours;
    private List<Seller> sellers = new List<Seller>();

    public string StoreName { get => storeName; set => storeName = value; }
    public string Location { get => location; set => location = value; }

    public Store(string storeName, string location, string storeHours)
    {
        this.storeName = storeName;
        this.location = location;
        this.storeHours = storeHours;
    }

    public void AddSeller(Seller seller) { sellers.Add(seller); }

    public void ListProducts()
    {
        Console.WriteLine($"Товары в '{storeName}'");
        foreach (var s in sellers)
            foreach (var p in s.Products)
                Console.WriteLine($"Продавец: {s.Name} → {p.GetProductInfo()}");
    }

    public string GetStoreInfo() => $"{storeName} | {location} | Часы: {storeHours}";
}


var product = new Product("Ноутбук", 75000m, 10, "Электроника", "Игровой ноутбук");
var seller = new Seller("Алябин П.С.", "EMP-001", 60000m, "albpets@shop.ru");
var store = new Store("DNS", "ул. Ленина, 5", "09:00–21:00");

seller.AddProduct(product);
store.AddSeller(seller);

Console.WriteLine(store.GetStoreInfo());
Console.WriteLine(seller.GetSellerInfo());
Console.WriteLine(product.GetProductInfo());
Console.WriteLine();

store.ListProducts();
Console.WriteLine();

seller.SellProduct(product, 3);
Console.WriteLine($"После продажи: {product.GetProductInfo()}");
Console.WriteLine($"Общая стоимость остатка: {product.GetTotalPrice()}₽");
```
