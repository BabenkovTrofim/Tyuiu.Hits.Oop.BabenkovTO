```
using System;
using System.Collections.Generic;

public interface IInstallable
{
    void Install();
}
public abstract class Component
{
    public string Name { get; set; }
    public string Manufacturer { get; set; }
    public double Price { get; set; }

    public Component(string name, string manufacturer, double price)
    {
        Name = name;
        Manufacturer = manufacturer;
        Price = price;
    }

    public virtual void GetInfo()
    {
        Console.WriteLine(ToString());
    }

    public override string ToString()
    {
        return $"|НАЗВАНИЕ КОМПОНЕНТА: {Name}|ПРОИЗВОДИТЕЛЬ: {Manufacturer} |СТОИМОСТЬ: {Price}|";
    }
}

public class CPU : Component, IInstallable
{
    public int Cores { get; set; }
    public string Socket { get; set; }

    public CPU(string name, string manufacturer, double price, int cores, string socket) : base(name, manufacturer, price)
    {
        Cores = cores;
        Socket = socket;
    }

    public static bool operator ==(CPU left, CPU right)
    {

        if (ReferenceEquals(left, right))
            return true;

        if (left is null || right is null)
            return false;
        
        return left.Name == right.Name
            && left.Cores == right.Cores
            && left.Socket == right.Socket;
    }

    public static bool operator !=(CPU left, CPU right)
    {
        return !(left == right);
    }

    public override bool Equals(object obj)
    {
        return obj is CPU other && this == other;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Cores, Socket);
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"|КОЛИЧЕСТВО ЯДЕР: {Cores} | ТИП СОКЕТА: {Socket}|");
    }
    public void Install()
    {
        Console.WriteLine("Процессор установлен!");
    }
}

public class RAM : Component, IInstallable
{
    public int Capacity { get; set; }
    public string MemoryType { get; set; }

    public RAM(string name, string manufacturer, double price, int capacity, string memoryType) : base(name, manufacturer, price)
    {
        Capacity = capacity;
        MemoryType = memoryType;
    }

    public static bool operator ==(RAM left, RAM right)
    {

        if (ReferenceEquals(left, right))
            return true;

        if (left is null || right is null)
            return false;
        
        return left.Name == right.Name
            && left.Capacity == right.Capacity
            && left.MemoryType == right.MemoryType;
    }

    public static bool operator !=(RAM left, RAM right)
    {
        return !(left == right);
    }

    public override bool Equals(object obj)
    {
        return obj is RAM other && this == other;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Capacity, MemoryType);
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"|ОБЪЕМ, ГБ: {Capacity} | ТИП ПАМЯТИ: {MemoryType}|");
    }
    public void Install()
    {
        Console.WriteLine("Оперативная память установлена!");
    }
}

public class GPU : Component, IInstallable
{
    public int VideoMemory { get; set; }
    public string Chipset { get; set; }

    public GPU(string name, string manufacturer, double price, int videoMemory, string chipset) : base(name, manufacturer, price)
    {
        VideoMemory = videoMemory;
        Chipset = chipset;
    }

    public static bool operator ==(GPU left, GPU right)
    {

        if (ReferenceEquals(left, right))
            return true;

        if (left is null || right is null)
            return false;
        
        return left.Name == right.Name
            && left.VideoMemory == right.VideoMemory
            && left.Chipset == right.Chipset;
    }

    public static bool operator !=(GPU left, GPU right)
    {
        return !(left == right);
    }

    public override bool Equals(object obj)
    {
        return obj is GPU other && this == other;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, VideoMemory, Chipset);
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"|ОБЪЕМ ВИДЕОПАМЯТИ, ГБ: {VideoMemory} | ИСПОЛНЕНИЕ: {Chipset}|");
    }
    public void Install()
    {
        Console.WriteLine("Графический процессор установлен!");
    }
}

public class HDD : Component, IInstallable
{
    public int Capacity { get; set; }
    public string Interface { get; set; }

    public HDD(string name, string manufacturer, double price, int capacity, string interfaceType) : base(name, manufacturer, price)
    {
        Capacity = capacity;
        Interface = interfaceType;
    }

    public static bool operator ==(HDD left, HDD right)
    {

        if (ReferenceEquals(left, right))
            return true;

        if (left is null || right is null)
            return false;
        
        return left.Name == right.Name
            && left.Capacity == right.Capacity
            && left.Interface == right.Interface;
    }

    public static bool operator !=(HDD left, HDD right)
    {
        return !(left == right);
    }

    public override bool Equals(object obj)
    {
        return obj is HDD other && this == other;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Capacity, Interface);
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"|ОБЪЕМ: {Capacity} | ИНТЕРФЕЙС ПОДКЛЮЧЕНИЯ: {Interface}|");
    }
    public void Install()
    {
        Console.WriteLine("Жесткий диск установлен!");
    }
}

public class ComponentCollection<T> where T : Component
{
    private List<T> _components = new List<T>();

    public int Count => _components.Count;

    public void Add(T component)
    {
        _components.Add(component);
    }

    public bool Remove(T component)
    {
        return _components.Remove(component);
    }

    public void DisplayComponents()
    {
        foreach (var component in _components)
        {
            component.GetInfo();    
        }
    }

    public double GetTotalPrice()
    {
        double total = 0;
        foreach (var component in _components)
        {
            total += component.Price;
        }
        return total;
    }
}

public class Computer
{
    private ComponentCollection<Component> _components = new ComponentCollection<Component>();

    public string Name { get; }

    public Computer(string name)
    {
        Name = name;
    }

    public void AddComponent<T>(T component) where T : Component, IInstallable
    {
        component.Install();
        _components.Add(component);
    }

    public bool RemoveComponent<T>(T component) where T : Component
    {
        bool removed = _components.Remove(component);

        Console.WriteLine(removed
            ? $"Удалён компонент: {component.Name}"
            : $"Компонент {component.Name} не найден в компьютере");

        return removed;
    }

    public double GetTotalPrice()
    {
        return _components.GetTotalPrice();
    }

    public void DisplayComputer()
    {
        Console.WriteLine($"=== Компьютер \"{Name}\" ===");
        _components.DisplayComponents();
        Console.WriteLine($"Итого: {GetTotalPrice():N0}");
    }
}

var cpu1 = new CPU("Core i5-12400", "Intel", 15000, 6, "LGA1700");
var cpu2 = new CPU("Core i5-12400", "Intel", 16500, 6, "LGA1700"); 
var cpu3 = new CPU("Ryzen 7 5800X", "AMD", 25000, 8, "AM4");

var ram1 = new RAM("Vengeance 16GB", "Corsair", 5000, 16, "DDR4");
var ram2 = new RAM("Fury 32GB", "Kingston", 9000, 32, "DDR5");

var gpu1 = new GPU("RTX 3060", "NVIDIA", 35000, 12, "GeForce RTX 3060");
var gpu2 = new GPU("RX 6600", "AMD", 24000, 8, "Radeon RX 6600");

var hdd1 = new HDD("Barracuda 2TB", "Seagate", 4500, 2000, "SATA III");
var hdd2 = new HDD("Blue 1TB", "Western Digital", 3200, 1000, "SATA III");

Console.WriteLine("===== 1. ПОЛИМОРФИЗМ: GetInfo() =====");
Component[] allComponents = { cpu1, ram1, gpu1, hdd1 };
foreach (Component component in allComponents)
{
    component.GetInfo();   
    Console.WriteLine();
}


Console.WriteLine("===== 2. СРАВНЕНИЕ КОМПОНЕНТОВ =====");
Console.WriteLine($"cpu1 == cpu2: {cpu1 == cpu2}");   
Console.WriteLine($"cpu1 != cpu3: {cpu1 != cpu3}");   
Console.WriteLine($"cpu1.Equals(cpu2): {cpu1.Equals(cpu2)}");
Console.WriteLine();

Console.WriteLine("===== 3. ComponentCollection<CPU> =====");
var cpuCollection = new ComponentCollection<CPU>();
cpuCollection.Add(cpu1);
cpuCollection.Add(cpu3);
cpuCollection.DisplayComponents();
Console.WriteLine($"Процессоров: {cpuCollection.Count}, сумма: {cpuCollection.GetTotalPrice()}");
Console.WriteLine();

Console.WriteLine("===== 4. СБОРКА КОМПЬЮТЕРА =====");
var gamingPc = new Computer("Игровой ПК");
gamingPc.AddComponent(cpu3);
gamingPc.AddComponent(ram2);
gamingPc.AddComponent(gpu1);
gamingPc.AddComponent(hdd1);
gamingPc.AddComponent(hdd2);  
Console.WriteLine();

gamingPc.DisplayComputer();
Console.WriteLine();

Console.WriteLine("===== 5. УДАЛЕНИЕ КОМПОНЕНТОВ =====");
gamingPc.RemoveComponent(hdd2);     
gamingPc.RemoveComponent(gpu2);    

Console.WriteLine($"Стоимость после удаления: {gamingPc.GetTotalPrice()}");
Console.WriteLine();
gamingPc.DisplayComputer();
Console.WriteLine();
```
