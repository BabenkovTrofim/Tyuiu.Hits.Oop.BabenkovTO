```
using System;
using System.Collections.Generic;
using System.Linq;

public delegate void LowStockEventHandler(Product product, int remainingStock);

Console.WriteLine("=== 1. Демонстрация событий и делегатов (Low Stock Alert) ===");
Product shirt = new Clothing("Футболка Basic", 1500, "Kappa", "M", "Хлопок", "Мужской", true);
shirt.Restock(3);

shirt.OnLowStock += WarnAboutLowStock;

shirt.Sell(2);

Console.WriteLine("\n=== 2. Новые методы и коллекции в базовом классе Product ===");
shirt.AddReview("Отличная ткань, рекомендую!");
shirt.AddReview("Размер маломерит.");
Console.WriteLine($"Отзывы о '{shirt.Name}': {string.Join(" | ", shirt.Reviews)}");
Console.WriteLine($"Стоимость остатков: {shirt.GetTotalStockValue():F2} руб.");

Console.WriteLine("\n=== 3. Новые методы и коллекции в Electronics ===");
Electronics tv = new Electronics("Телевизор LG", 45000, "LG", 24, "Белый", true, 0);
tv.PowerWatts = 120;
tv.AddFeature("Smart TV");
tv.AddFeature("4K HDR");
tv.RunDiagnostics();
Console.WriteLine($"Годовая стоимость энергии: {tv.EstimateYearlyEnergyCost(6.5, 4):F2} руб.");

Console.WriteLine("\n=== 4. Новые методы и коллекции в Clothing ===");
Clothing jacket = new Clothing("Куртка Зимняя", 12000, "Nike", "L", "Полиэстер", "Унисекс", false);
jacket.Season = "Зима";
jacket.AddSize("M");
jacket.AddSize("L");
jacket.AddSize("XL");
Console.WriteLine($"Доступные размеры: {string.Join(", ", jacket.AvailableSizes)}");
jacket.ApplySeasonalDiscount("Зима");

Console.WriteLine("\n=== 5. Новые методы и коллекции в Smartphone (Dictionary) ===");
Smartphone phone = new Smartphone("Galaxy S24", 80000, "Samsung", 12, "Android 14", "Серый");
phone.InstallApp("Telegram", 150);
phone.InstallApp("Spotify", 80);
phone.InstallApp("Genshin Impact", 15000);
Console.WriteLine($"Занято памяти: {phone.GetTotalStorageUsed()} МБ из {phone.StorageGb * 1024} МБ");

Console.WriteLine("\n=== 6. Новые методы и коллекции в SmartWatch (Queue) ===");
var memoryLogger = new MemoryLogger();
var watch = new SmartWatch("Galaxy Watch 6", 25000, "Samsung", 12, memoryLogger);
watch.ReceiveNotification("Telegram", "Новое сообщение от Алисы");
watch.ReceiveNotification("Calendar", "Встреча через 15 минут");
watch.ReceiveNotification("Health", "Цель по шагам достигнута!");
watch.ShowNotifications();

Console.WriteLine("\n=== 7. Перекрытие методов (Полиморфизм) ===");
Product[] mixedProducts =
{
    new Electronics("Наушники", 5000, "Sony", 12, "Черный"),
    new Clothing("Джинсы", 3000, "Levi's", "32", "Деним"),
    new Smartphone("iPhone 15", 90000, "Apple", 12, "iOS 17", "Титан")
};
foreach (var prod in mixedProducts) prod.PrintDetails();

Console.WriteLine("\n=== 8. Generic класс ProductCatalog<T> ===");
var phoneCatalog = new ProductCatalog<Smartphone>();
phoneCatalog.AddItem(new Smartphone("Pixel 8", 70000, "Google", 12, "Android 14", "Черный"));
phoneCatalog.AddItem(phone);
phoneCatalog.DisplayAll();
phoneCatalog.ApplyBulkDiscount(5);

static void WarnAboutLowStock(Product product, int remainingStock)
{
    Console.WriteLine($"[ВНИМАНИЕ СОБЫТИЯ] Товар '{product.Name}' заканчивается! Осталось всего {remainingStock} шт. Срочно закажите поставку.");
}

public interface IWaterproof { void CheckWaterResistance(); }
public interface IHealthTracker { void MeasureHeartRate(); }
public interface ILogger { void Log(string message); }

public class ConsoleLogger : ILogger
{
    public void Log(string message) => Console.WriteLine(message);
}

public class MemoryLogger : ILogger
{
    public List<string> Messages { get; } = new List<string>();
    public void Log(string message) => Messages.Add(message);
}

public class Product
{
    private string name;
    private double price;
    private string manufacturer;
    private int stockQuantity;
    private DateTime manufactureDate;
    private double weightKg;

    public int MinStockThreshold { get; set; } = 5;
    public List<string> Reviews { get; } = new List<string>();

    public event LowStockEventHandler OnLowStock;

    public string Name { get => name; set => name = string.IsNullOrWhiteSpace(value) ? "Неизвестно" : value; }
    public double Price { get => price; set => price = value < 0 ? 0 : value; }
    public string Manufacturer { get => manufacturer; set => manufacturer = string.IsNullOrWhiteSpace(value) ? "Неизвестно" : value; }
    public int StockQuantity { get => stockQuantity; set => stockQuantity = value < 0 ? 0 : value; }
    public DateTime ManufactureDate { get => manufactureDate; set => manufactureDate = value; }
    public double WeightKg { get => weightKg; set => weightKg = value < 0 ? 0 : value; }
    public bool IsActive { get; private set; } = true;

    public Product(string name, double price, string manufacturer, DateTime? manufactureDate = null)
    {
        Name = name;
        Price = price;
        Manufacturer = manufacturer;
        ManufactureDate = manufactureDate ?? DateTime.Now;
    }

    public void Restock(int amount) { StockQuantity += amount; Console.WriteLine($"[Склад] '{Name}' пополнен на {amount} шт. Остаток: {StockQuantity}"); }

    public virtual void Sell(int quantity)
    {
        if (StockQuantity >= quantity)
        {
            StockQuantity -= quantity;
            Console.WriteLine($"[Продажа] Продано {quantity} шт. '{Name}'. Остаток: {StockQuantity}");

            if (StockQuantity <= MinStockThreshold)
            {
                OnLowStock?.Invoke(this, StockQuantity);
            }
        }
        else
        {
            Console.WriteLine($"[Ошибка] Недостаточно товара '{Name}' для продажи.");
        }
    }

    public void AddReview(string review) => Reviews.Add(review);
    public double GetTotalStockValue() => Price * StockQuantity;

    public virtual void PrintDetails()
    {
        Console.WriteLine($"[Товар] {Name} | Производитель: {Manufacturer} | Рейтинг: {Reviews.Count} отзывов");
    }
}

public class Electronics : Product
{
    private int warrantyPeriod;
    private string color;
    private bool hasWifi;
    private double powerWatts;

    public List<string> SupportedFeatures { get; } = new List<string>();
    public bool IsOverheating { get; private set; }

    public int WarrantyPeriod { get => warrantyPeriod; set => warrantyPeriod = value < 0 ? 0 : value; }
    public string Color { get => color; set => color = string.IsNullOrWhiteSpace(value) ? "Черный" : value; }
    public bool HasWifi { get => hasWifi; set => hasWifi = value; }
    public double PowerWatts { get => powerWatts; set => powerWatts = value < 0 ? 0 : value; }

    public Electronics(string name, double price, string manufacturer, int warrantyPeriod, string color = "Черный", bool hasWifi = false, double powerWatts = 0)
        : base(name, price, manufacturer)
    {
        WarrantyPeriod = warrantyPeriod;
        Color = color;
        HasWifi = hasWifi;
        PowerWatts = powerWatts;
    }

    public void AddFeature(string feature) => SupportedFeatures.Add(feature);

    public void RunDiagnostics()
    {
        IsOverheating = PowerWatts > 200;
        Console.WriteLine($"[Диагностика] {Name}: Перегрев = {IsOverheating}, Активные фичи: {string.Join(", ", SupportedFeatures)}");
    }

    public double EstimateYearlyEnergyCost(double tariffPerKwh, double hoursPerDay)
        => PowerWatts * hoursPerDay * 365 / 1000.0 * tariffPerKwh;

    public override void PrintDetails()
    {
        base.PrintDetails();
        Console.WriteLine($"  -> Электроника: Цвет={Color}, Wi-Fi={HasWifi}, Гарантия={WarrantyPeriod} мес.");
    }
}

public class Clothing : Product
{
    private string size;
    private string material;
    private string gender;
    private bool isWashable;

    public List<string> AvailableSizes { get; } = new List<string>();
    public double SeasonalDiscountRate { get; private set; } = 0.0;
    public string Season { get; set; } = "Всесезонная";

    public string Size { get => size; set => size = string.IsNullOrWhiteSpace(value) ? "Не указан" : value; }
    public string Material { get => material; set => material = string.IsNullOrWhiteSpace(value) ? "Не указан" : value; }
    public string Gender { get => gender; set => gender = string.IsNullOrWhiteSpace(value) ? "Унисекс" : value; }
    public bool IsWashable { get => isWashable; set => isWashable = value; }

    public Clothing(string name, double price, string manufacturer, string size, string material = "Хлопок", string gender = "Унисекс", bool isWashable = true)
        : base(name, price, manufacturer)
    {
        Size = size;
        Material = material;
        Gender = gender;
        IsWashable = isWashable;
    }

    public void AddSize(string size) => AvailableSizes.Add(size);

    public void ApplySeasonalDiscount(string targetSeason)
    {
        if (Season == "Всесезонная" || string.Equals(Season, targetSeason, StringComparison.OrdinalIgnoreCase))
        {
            SeasonalDiscountRate = 0.15;
            Price = Price * (1 - SeasonalDiscountRate);
            Console.WriteLine($"[Сезонная скидка] На '{Name}' применена скидка 15% для сезона '{targetSeason}'. Новая цена: {Price:F2}");
        }
        else
        {
            Console.WriteLine($"[Сезонная скидка] '{Name}' не подходит для сезона '{targetSeason}'. Скидка не применена.");
        }
    }

    public override void PrintDetails()
    {
        base.PrintDetails();
        Console.WriteLine($"  -> Одежда: Размер={Size}, Материал={Material}, Пол={Gender}, Сезон={Season}");
    }
}

public class Smartphone : Electronics
{
    private string os;

    public int StorageGb { get; set; } = 128;
    public Dictionary<string, int> InstalledApps { get; } = new Dictionary<string, int>();

    public string Os { get => os; set => os = string.IsNullOrWhiteSpace(value) ? "Unknown OS" : value; }

    public Smartphone(string name, double price, string manufacturer, int warrantyPeriod, string os, string color = "Черный")
        : base(name, price, manufacturer, warrantyPeriod, color, hasWifi: true, powerWatts: 15)
    {
        Os = os;
    }

    public void InstallApp(string appName, int sizeMb)
    {
        int totalUsedMb = GetTotalStorageUsed();
        int totalStorageMb = StorageGb * 1024;

        if (totalStorageMb >= totalUsedMb + sizeMb)
        {
            InstalledApps[appName] = sizeMb;
            Console.WriteLine($"[Установка] {appName} ({sizeMb} МБ) успешно установлен на {Name}.");
        }
        else
        {
            Console.WriteLine($"[Ошибка] Недостаточно памяти для установки {appName}. Свободно: {totalStorageMb - totalUsedMb} МБ.");
        }
    }

    public int GetTotalStorageUsed() => InstalledApps.Values.Sum();

    public override void PrintDetails()
    {
        base.PrintDetails();
        Console.WriteLine($"    -> Смартфон: ОС={Os}, Память={StorageGb} ГБ, Приложений: {InstalledApps.Count}");
    }
}

public class SmartWatch : Electronics, IWaterproof, IHealthTracker
{
    private readonly ILogger _logger;
    public int StepCount { get; private set; }

    public Queue<string> NotificationQueue { get; } = new Queue<string>();

    public SmartWatch(string name, double price, string manufacturer, int warrantyPeriod, ILogger logger = null)
        : base(name, price, manufacturer, warrantyPeriod, color: "Черный", hasWifi: true, powerWatts: 5)
    {
        _logger = logger ?? new ConsoleLogger();
    }

    void IWaterproof.CheckWaterResistance()
        => _logger.Log($"[Тест] {Name} проверен на водонепроницаемость (5 ATM). Утечек нет.");

    void IHealthTracker.MeasureHeartRate()
        => _logger.Log($"[Здоровье] {Name} измеряет пульс: 72 уд/мин.");

    public void AddSteps(int steps)
    {
        StepCount += steps;
        _logger.Log($"[Шаги] {Name}: всего {StepCount} шагов.");
    }

    public void ReceiveNotification(string sender, string message)
    {
        string note = $"[{sender}] {message}";
        NotificationQueue.Enqueue(note);
        _logger.Log($"[Уведомление] Получено: {note}");

        if (NotificationQueue.Count > 5)
        {
            NotificationQueue.Dequeue();
        }
    }

    public void ShowNotifications()
    {
        string notes = NotificationQueue.Count > 0 ? string.Join(" | ", NotificationQueue) : "Нет новых уведомлений";
        Console.WriteLine($"[Экран {Name}] Уведомления: {notes}");
    }
}

public class ProductCatalog<T> where T : Product
{
    private List<T> _items = new List<T>();

    public void AddItem(T item)
    {
        _items.Add(item);
        Console.WriteLine($"[Каталог] Добавлен: {item.Name}");
    }

    public void DisplayAll()
    {
        Console.WriteLine("\n--- Содержимое каталога ---");
        foreach (var item in _items) item.PrintDetails();
        Console.WriteLine("---------------------------\n");
    }

    public void ApplyBulkDiscount(double percent)
    {
        Console.WriteLine($"\n[Каталог] Применяем оптовую скидку {percent}% ко всем товарам...");
        foreach (var item in _items)
        {
            Console.WriteLine($" -> Скидка для {item.Name} обработана.");
        }
    }
}
```
