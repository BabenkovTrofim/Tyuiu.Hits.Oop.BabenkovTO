```
public interface IWaterproof
{
    void CheckWaterResistance();
}

public interface IHealthTracker
{
    void MeasureHeartRate();
}

public class Product
{
    private string name;
    private double price;
    private string manufacturer;

    private string sku;     // артикул
    private int stockQuantity; // количество на складе
    private string category;   // категория товара

    public string Name
    {
        get { return name; }
        set { name = string.IsNullOrWhiteSpace(value) ? "Неизвестно" : value; }
    }

    public double Price
    {
        get { return price; }
        set { price = value < 0 ? 0 : value; }
    }

    public string Manufacturer
    {
        get { return manufacturer; }
        set { manufacturer = string.IsNullOrWhiteSpace(value) ? "Неизвестно" : value; }
    }

    public string Sku { get => sku; set => sku = string.IsNullOrWhiteSpace(value) ? "N/A" : value; }
    public int StockQuantity { get => stockQuantity; set => stockQuantity = value < 0 ? 0 : value; }
    public string Category { get => category; set => category = string.IsNullOrWhiteSpace(value) ? "Общее" : value; }

    public Product(string name, double price, string manufacturer, string sku = "N/A", int stockQuantity = 0, string category = "Общее")
    {
        Name = name;
        Price = price;
        Manufacturer = manufacturer;
        Sku = sku;
        StockQuantity = stockQuantity;
        Category = category;
    }

    public void Restock(int amount)
    {
        StockQuantity += amount;
        Console.WriteLine($"[Склад] Товар '{Name}' пополнен на {amount} шт. Остаток: {StockQuantity}");
    }

    public bool IsInStock()
    {
        return StockQuantity > 0;
    }

    public void PrintLabel()
    {
        Console.WriteLine($"--- Этикетка ---\nАртикул: {Sku} | Категория: {Category}\n{GetInfo()}\n----------------");
    }

    public virtual string GetInfo()
    {
        return $"Название: {Name}, Цена: {Price}, Производитель: {Manufacturer}";
    }

    public virtual void Discount(double percent)
    {
        Price = Price - (Price * percent / 100);
        Console.WriteLine($"Скидка {percent}% применена. Новая цена: {Price}");
    }

    public virtual void Display()
    {
        Console.WriteLine(GetInfo());
    }

    public virtual void BundleWith(Product otherProduct)
    {
        if (otherProduct != null)
        {
            this.Discount(5);
            otherProduct.Discount(5);
        }
    }
}

public class Electronics : Product
{
    private int warrantyPeriod;

    private string energyClass;  // Класс энергоэффективности
    private double weight;       // Вес в кг
    private bool isSmart;        // Является ли умным устройством

    public int WarrantyPeriod
    {
        get { return warrantyPeriod; }
        set { warrantyPeriod = value < 0 ? 0 : value; }
    }

    public string EnergyClass { get => energyClass; set => energyClass = string.IsNullOrWhiteSpace(value) ? "A" : value; }
    public double Weight { get => weight; set => weight = value < 0 ? 0 : value; }
    public bool IsSmart { get => isSmart; set => isSmart = value; }

    public Electronics(string name, double price, string manufacturer, int warrantyPeriod, 
                       string energyClass = "A", double weight = 0.5, bool isSmart = false)
        : base(name, price, manufacturer)
    {
        WarrantyPeriod = warrantyPeriod;
        EnergyClass = energyClass;
        Weight = weight;
        IsSmart = isSmart;
    }

    public void TurnOn()
    {
        Console.WriteLine($"[Устройство] {Name} включено.");
    }

    public void TurnOff()
    {
        Console.WriteLine($"[Устройство] {Name} выключено.");
    }

    public string GetEnergyInfo()
    {
        return $"Класс энергоэффективности: {EnergyClass}, Вес: {Weight} кг";
    }

    public override string GetInfo()
    {
        return $"Название: {Name}, Цена: {Price}, Производитель: {Manufacturer}, Гарантия: {WarrantyPeriod} м.";
    }

    public override void Discount(double percent)
    {
        if (WarrantyPeriod > 12)
        {
            percent = percent + 5;
            Console.WriteLine($"Гарантия {WarrantyPeriod} мес. — дополнительная скидка 5%!");
        }

        Price = Price - (Price * percent / 100);
        Console.WriteLine($"Название: {Name}. Скидка {percent}% применена. Новая цена: {Price}");
    }
}

public class Clothing : Product
{
    private string size;

    public string Size
    {
        get { return size; }
        set { size = string.IsNullOrWhiteSpace(value) ? "Не указан" : value; }
    }

    public Clothing(string name, double price, string manufacturer, string size)
        : base(name, price, manufacturer)
    {
        Size = size;
    }

    public override void Display()
    {
        Console.WriteLine($"Название: {Name}, Цена: {Price}, Производитель: {Manufacturer}, Размер: {Size}");
    }
}

public class Smartphone : Electronics
{
    private string os;

    public string Os { get => os; set => os = string.IsNullOrWhiteSpace(value) ? "Unknown OS" : value; }

    public Smartphone(string name, double price, string manufacturer, int warrantyPeriod, string os)
        : base(name, price, manufacturer, warrantyPeriod, isSmart: true)
    {
        Os = os;
    }

    public void MakeCall(string number)
    {
        Console.WriteLine($"[Звонок] Абонент {number} вызывается с {Name} ({Os})...");
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", ОС: {Os}";
    }
}


public class SmartWatch : Electronics, IWaterproof, IHealthTracker
{
    public SmartWatch(string name, double price, string manufacturer, int warrantyPeriod)
        : base(name, price, manufacturer, warrantyPeriod, weight: 0.05, isSmart: true) { }

    public void CheckWaterResistance()
    {
        Console.WriteLine($"[Тест] {Name} проверен на водонепроницаемость (5 ATM). Утечек нет.");
    }

    public void MeasureHeartRate()
    {
        Console.WriteLine($"[Здоровье] {Name} измеряет пульс: 72 уд/мин.");
    }
}

public class ShoppingCart
{
    private List<Product> items = new List<Product>();

    public void Add(Product product)
    {
        items.Add(product);
    }

    public void ApplyCartDiscount(double percent)
    {
        foreach (var item in items)
        {
            item.Discount(percent);
        }
    }

    public double GetTotal()
    {
        double total = 0;
        foreach (var item in items)
        {
            total += item.Price;
        }
        return total;
    }
}


Product phone = new Electronics("Смартфон", 30000, "Samsung", 24, "A+", 0.2, true);
phone.Restock(10); // Новый метод
phone.PrintLabel(); // Новый метод

Console.WriteLine("\n--- Сложное (многоуровневое) наследование (Smartphone) ---");
Smartphone myPhone = new Smartphone("iPhone 15", 80000, "Apple", 12, "iOS 17");
myPhone.TurnOn();       // Метод из Electronics
myPhone.MakeCall("112"); // Метод из Smartphone
myPhone.Restock(5);     // Метод из Product

Console.WriteLine("\n--- Множественное наследование (SmartWatch + Интерфейсы) ---");
SmartWatch watch = new SmartWatch("Galaxy Watch", 25000, "Samsung", 12);
watch.CheckWaterResistance(); // Метод из интерфейса IWaterproof
watch.MeasureHeartRate();     // Метод из интерфейса IHealthTracker
watch.TurnOn();               // Метод из класса Electronics

Console.WriteLine("\n--- Работа с корзиной ---");
Product notebook = new Electronics("Ноутбук", 59999, "ASUS", 12);
phone.BundleWith(notebook);

ShoppingCart cart = new ShoppingCart();
cart.Add(phone);
cart.Add(notebook);
cart.Add(new Clothing("Футболка", 2500, "Kappa", "M"));
cart.Add(myPhone);
cart.Add(watch);

cart.ApplyCartDiscount(3);

Console.WriteLine($"\nИтоговая сумма корзины: {cart.GetTotal()}");
```
