<h4> Описание задачи:</h4>
Создать базовый класс Product в C#, который будет представлять информацию о
продуктах. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.


<h4>Требования к базовому классу Product:</h4>
• Атрибуты: Название (Name), Цена (Price), Производитель (Manufacturer).
• Методы:
o GetInfo(): метод для получения информации о продукте в виде строки.
o Discount(): метод для применения скидки к цене продукта.
o Display(): метод для отображения информации о продукте на экране.

<h4>Требования к производным классам:</h4>

1. Электроника (Electronics): Должен содержать дополнительные атрибуты,
такие как Гарантийный срок (WarrantyPeriod). Метод Discount() должен быть
переопределен для добавления логики учета гарантийного срока при
применении скидки.
2. Одежда (Clothing): Должен содержать дополнительные атрибуты, такие как
Размер (Size). Метод Display() должен быть переопределен для добавления
информации о размере при отображении информации о продукте.
3. Книги (Books) (если требуется третий класс): Должен содержать
дополнительные атрибуты, такие как Автор (Author). Метод GetInfo() должен
быть переопределен для включения информации об авторе в описании
продукта.


<h2 style="color:DodgerBlue">Реализация:</h2>

```
public class Product
{
    public string Name { get; set; }
    public double Price { get; set; }
    public string Manufacturer { get; set; }

    public Product(string name, double price, string manufacturer)
    {
        Name = name;
        Price = price;
        Manufacturer = manufacturer;
    }

    // получение инфы о продукте
    public virtual string GetInfo()
    {
        return $"Название: {Name}, Цена: {Price}, Производитель: {Manufacturer}";
    }

    // применение скидки
    public virtual void Discount(double percent)
    {
        Price = Price - (Price * percent / 100);
        Console.WriteLine($"Скидка {percent}% применена. Новая цена: {Price}");
    }

    // вывод инфы на экран
    public virtual void Display()
    {
        Console.WriteLine(GetInfo());
    }
}

public class Electronics : Product
{
    // атрибут добавления гарантии
    public int WarrantyPeriod { get; set; } // в месяцах

    public Electronics(string name, double price, string manufacturer, int warrantyPeriod)
        : base(name, price, manufacturer)
    {
        WarrantyPeriod = warrantyPeriod;
    }
    public override string GetInfo()
    {
        return $"Название: {Name}, Цена: {Price}, Производитель: {Manufacturer}, Гарантия: {WarrantyPeriod} м.";
    }

    // скидка с учетом срока гарантии(гарантия >12 месяцкв - доп скидка +5%)
    public override void Discount(double percent)
    {
        if (WarrantyPeriod > 12)
        {
            percent = percent + 5;
            Console.WriteLine($"Гарантия {WarrantyPeriod} мес. — дополнительная скидка 5%!");
        }

        Price = Price - (Price * percent / 100);
        Console.WriteLine($"Название: {Name}. Скидка {percent}% применена. Новая цена: {Price}");
    }
}

public class Clothing : Product
{
    public string Size { get; set; } // например, "M", "L", "XL"

    public Clothing(string name, double price, string manufacturer, string size)
        : base(name, price, manufacturer)
    {
        Size = size;
    }

    public override void Display()
    {
        Console.WriteLine($"Название: {Name}, Цена: {Price}, Производитель: {Manufacturer}, Размер: {Size}");
    }
}
// ЭЛЕКТРОНИКА
Console.WriteLine("ЭЛЕКТРОНИКА:");
Product phone = new Electronics("Смартфон", 30000, "Samsung", 24);
Product notebook = new Electronics("Ноутбук", 59999, "ASUS", 12);
Product speakers = new Electronics("Динамики", 19999, "Apocalypse", 12);


Product[] electronic = new Product[] { phone, notebook, speakers};
foreach (Product electronic in electronic)
{
    electronic.Display();       
}
Console.WriteLine();
Console.WriteLine("Информация с учетом скидок и акций:");
phone.Discount(0);
notebook.Discount(10);
speakers.Discount(15);
Console.WriteLine();
foreach (Product electronic in electronic)
{
    Console.WriteLine(electronic.GetInfo());
}
Console.WriteLine();

// ОДЕЖДА
Console.WriteLine("ОДЕЖДА:");
Product shirt = new Clothing("Футболка", 2500, "Kappa", "M");
Product trousers = new Clothing("Штаны спортивные", 3000, "Adidas", "L");
Product sneakers = new Clothing("Кроссовки", 7000, "Nike", "41");


Product[] clothing = new Product[] { shirt, trousers, sneakers};
foreach (Product clothing in clothing)
{
    clothing.Display();       
}
Console.WriteLine();
Console.WriteLine("Информация с учетом скидок и акций:");

shirt.Discount(5);
trousers.Discount(10);
sneakers.Discount(13);

Console.WriteLine();
foreach (Product clothing in clothing)
{
    Console.WriteLine(clothing.GetInfo());
}
```
