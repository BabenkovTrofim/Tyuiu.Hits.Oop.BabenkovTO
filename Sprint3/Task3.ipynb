```
using System;
using System.Collections.Generic;
using System.Linq;

public interface IWaterproof
{
    void CheckWaterResistance();
}

public interface IHealthTracker
{
    void MeasureHeartRate();
}

public class Product
{
    
    private string name;
    private double price;
    private string manufacturer;
    private string sku;
    private int stockQuantity;
    private string category;

    // 3 атрибута
    private double rating;
    private string countryOfOrigin;
    private DateTime manufactureDate;

  
    public string Name { get => name; set => name = string.IsNullOrWhiteSpace(value) ? "Неизвестно" : value; }
    public double Price { get => price; set => price = value < 0 ? 0 : value; }
    public string Manufacturer { get => manufacturer; set => manufacturer = string.IsNullOrWhiteSpace(value) ? "Неизвестно" : value; }
    public string Sku { get => sku; set => sku = string.IsNullOrWhiteSpace(value) ? "N/A" : value; }
    public int StockQuantity { get => stockQuantity; set => stockQuantity = value < 0 ? 0 : value; }
    public string Category { get => category; set => category = string.IsNullOrWhiteSpace(value) ? "Общее" : value; }
    
    // cвойства для новых атрибутов
    public double Rating { get => rating; set => rating = value < 0 ? 0 : (value > 5 ? 5 : value); }
    public string CountryOfOrigin { get => countryOfOrigin; set => countryOfOrigin = string.IsNullOrWhiteSpace(value) ? "Не указана" : value; }
    public DateTime ManufactureDate { get => manufactureDate; set => manufactureDate = value; }

    public Product(string name, double price, string manufacturer, string countryOfOrigin = "Не указана", DateTime? manufactureDate = null)
    {
        Name = name;
        Price = price;
        Manufacturer = manufacturer;
        CountryOfOrigin = countryOfOrigin;
        ManufactureDate = manufactureDate ?? DateTime.Now;
        Rating = 0.0;
    }

    public void Restock(int amount) { StockQuantity += amount; Console.WriteLine($"[Склад] Товар '{Name}' пополнен на {amount} шт. Остаток: {StockQuantity}"); }
    public bool IsInStock() => StockQuantity > 0;
    
    public virtual void Discount(double percent)
    {
        Price = Price - (Price * percent / 100);
        Console.WriteLine($"[Скидка] {percent}% применена. Новая цена: {Price:F2}");
    }

    // новые методы 
    public void UpdateRating(double newRating)
    {
        Rating = newRating;
        Console.WriteLine($"[Рейтинг] Для '{Name}' установлен новый рейтинг: {Rating}/5");
    }

    public int GetAgeInMonths()
    {
        int months = (DateTime.Now.Year - ManufactureDate.Year) * 12 + DateTime.Now.Month - ManufactureDate.Month;
        return months < 0 ? 0 : months;
    }

    public virtual void PrintDetails()
    {
        Console.WriteLine($"[Товар] {Name} | Производитель: {Manufacturer} | Страна: {CountryOfOrigin} | Рейтинг: {Rating}");
    }

    public virtual void Discount(double percent, string promoCode)
    {
        if (promoCode == "SALE2024")
        {
            percent += 5; 
            Console.WriteLine($"[Промокод] '{promoCode}' активирован! Дополнительная скидка 5%.");
        }
        Discount(percent); 
    }
}

public class Electronics : Product
{
    private int warrantyPeriod;
    private string energyClass;
    private double weight;
    private bool isSmart;

    // 3 атрибута
    private string color;
    private bool hasWifi;
    private int batteryLifeMinutes;

    public int WarrantyPeriod { get => warrantyPeriod; set => warrantyPeriod = value < 0 ? 0 : value; }
    public string Color { get => color; set => color = string.IsNullOrWhiteSpace(value) ? "Черный" : value; }
    public bool HasWifi { get => hasWifi; set => hasWifi = value; }
    public int BatteryLifeMinutes { get => batteryLifeMinutes; set => batteryLifeMinutes = value < 0 ? 0 : value; }

    public Electronics(string name, double price, string manufacturer, int warrantyPeriod, string color = "Черный", bool hasWifi = false, int batteryLifeMinutes = 0)
        : base(name, price, manufacturer)
    {
        WarrantyPeriod = warrantyPeriod;
        Color = color;
        HasWifi = hasWifi;
        BatteryLifeMinutes = batteryLifeMinutes;
    }

    public void TurnOn()
    {
        Console.WriteLine($"[Устройство] {Name} включено в обычном режиме.");
    }

    public void TurnOn(bool silentMode)
    {
        string mode = silentMode ? "в тихом режиме" : "в обычном режиме";
        Console.WriteLine($"[Устройство] {Name} включено {mode}.");
    }

    // методы
    public void ConnectToWifi(string ssid)
    {
        if (HasWifi) Console.WriteLine($"[Wi-Fi] {Name} подключен к сети '{ssid}'.");
        else Console.WriteLine($"[Wi-Fi] {Name} не поддерживает Wi-Fi.");
    }

    public void ShowBatteryStatus()
    {
        Console.WriteLine($"[Батарея] {Name}: заряд рассчитан на {BatteryLifeMinutes} мин.");
    }

    public override void PrintDetails()
    {
        base.PrintDetails();
        Console.WriteLine($"  -> Электроника: Цвет={Color}, Wi-Fi={HasWifi}, Гарантия={WarrantyPeriod} мес.");
    }

    public override void Discount(double percent)
    {
        if (WarrantyPeriod > 12)
        {
            percent += 5;
            Console.WriteLine($"[Бонус] Длительная гарантия ({WarrantyPeriod} мес.) дает доп. скидку 5%!");
        }
        base.Discount(percent);
    }
}

public class Clothing : Product
{
    
    private string size;

    //3 атрибута
    private string material;
    private string gender;
    private bool isWashable;

    public string Size { get => size; set => size = string.IsNullOrWhiteSpace(value) ? "Не указан" : value; }
    public string Material { get => material; set => material = string.IsNullOrWhiteSpace(value) ? "Не указан" : value; }
    public string Gender { get => gender; set => gender = string.IsNullOrWhiteSpace(value) ? "Унисекс" : value; }
    public bool IsWashable { get => isWashable; set => isWashable = value; }

    public Clothing(string name, double price, string manufacturer, string size, string material = "Хлопок", string gender = "Унисекс", bool isWashable = true)
        : base(name, price, manufacturer)
    {
        Size = size;
        Material = material;
        Gender = gender;
        IsWashable = isWashable;
    }

    public bool CheckFit(string userSize)
    {
        bool fits = (Size == userSize);
        Console.WriteLine($"[Примерка] {Name} (размер {Size}): подходит пользователю с размером {userSize}? {fits}");
        return fits;
    }

    public bool CheckFit(string userSize, bool isStretchable)
    {
        if (isStretchable && Size != userSize)
        {
            Console.WriteLine($"[Примерка] {Name}: размер не совпадает, но ткань тянется, поэтому подойдет!");
            return true;
        }
        return CheckFit(userSize); 
    }

    public string GetCareInstructions()
    {
        return IsWashable ? "Можно стирать в машинке при 30°C." : "Только химчистка.";
    }

    public override void PrintDetails()
    {
        base.PrintDetails();
        Console.WriteLine($"  -> Одежда: Размер={Size}, Материал={Material}, Пол={Gender}");
    }
}

public class Smartphone : Electronics
{
    private string os;
    public string Os { get => os; set => os = string.IsNullOrWhiteSpace(value) ? "Unknown OS" : value; }

    public Smartphone(string name, double price, string manufacturer, int warrantyPeriod, string os, string color = "Черный")
        : base(name, price, manufacturer, warrantyPeriod, color, hasWifi: true, batteryLifeMinutes: 1440)
    {
        Os = os;
    }

    public void MakeCall(string number)
    {
        Console.WriteLine($"[Звонок] Абонент {number} вызывается с {Name} ({Os})...");
    }

    public override void PrintDetails()
    {
        base.PrintDetails();
        Console.WriteLine($"    -> Смартфон: ОС={Os}");
    }
}

public class SmartWatch : Electronics, IWaterproof, IHealthTracker
{
    public SmartWatch(string name, double price, string manufacturer, int warrantyPeriod)
        : base(name, price, manufacturer, warrantyPeriod, color: "Черный", hasWifi: true, batteryLifeMinutes: 2880) { }

    public void CheckWaterResistance()
    {
        Console.WriteLine($"[Тест] {Name} проверен на водонепроницаемость (5 ATM). Утечек нет.");
    }

    public void MeasureHeartRate()
    {
        Console.WriteLine($"[Здоровье] {Name} измеряет пульс: 72 уд/мин.");
    }
}


public class ProductCatalog<T> where T : Product
{
    private List<T> _items = new List<T>();

    public void AddItem(T item)
    {
        _items.Add(item);
        Console.WriteLine($"[Каталог] Добавлен: {item.Name}");
    }

    public void RemoveItem(T item)
    {
        if (_items.Remove(item))
        {
            Console.WriteLine($"[Каталог] Удален: {item.Name}");
        }
    }

    public void DisplayAll()
    {
        Console.WriteLine("\n--- Содержимое каталога ---");
        foreach (var item in _items)
        {
            item.PrintDetails();
        }
        Console.WriteLine("---------------------------\n");
    }

    public T GetMostExpensive()
    {
        return _items.OrderByDescending(i => i.Price).FirstOrDefault();
    }

    public void ApplyBulkDiscount(double percent)
    {
        Console.WriteLine($"\n[Каталог] Применяем оптовую скидку {percent}% ко всем товарам...");
        foreach (var item in _items)
        {
            item.Discount(percent);
        }
    }
}


Console.WriteLine("1. новые атрибутов и методов ");
Product shirt = new Clothing("Футболка Basic", 1500, "Kappa", "M", "Хлопок", "Мужской", true);
shirt.UpdateRating(4.5); 
Console.WriteLine($"Возраст товара на складе: {shirt.GetAgeInMonths()} мес.");
Console.WriteLine($"Инструкция по уходу: {((Clothing)shirt).GetCareInstructions()}"); 

Console.WriteLine("\n 2.перегрузка методов (Overloading)");
Electronics tv = new Electronics("Телевизор", 45000, "LG", 24, "Белый", true, 0);
tv.TurnOn(); 
tv.TurnOn(true); 

Console.WriteLine("\nПрименение скидки без промокода:");
shirt.Discount(10);

Console.WriteLine("\nПрименение скидки с промокодом:");
shirt.Discount(10, "SALE2024"); 

Console.WriteLine("\n3.перекрытие методов (Overriding / Полиморфизм)");
Product[] mixedProducts = new Product[]
{
    new Electronics("Наушники", 5000, "Sony", 12, "Черный"),
    new Clothing("Джинсы", 3000, "Levi's", "32", "Деним"),
    new Smartphone("Galaxy S24", 80000, "Samsung", 12, "Android 14", "Серый")
};

foreach (var prod in mixedProducts)
{
    prod.PrintDetails(); 
}

Console.WriteLine("\n4. Generic класс ProductCatalog<T> ");
ProductCatalog<Smartphone> phoneCatalog = new ProductCatalog<Smartphone>();
phoneCatalog.AddItem(new Smartphone("iPhone 15", 90000, "Apple", 12, "iOS 17", "Титан"));
phoneCatalog.AddItem(new Smartphone("Pixel 8", 70000, "Google", 12, "Android 14", "Черный"));


phoneCatalog.DisplayAll();

Smartphone mostExpensive = phoneCatalog.GetMostExpensive();
Console.WriteLine($"Самый дорогой смартфон в каталоге: {mostExpensive.Name} ({mostExpensive.Price} руб.)");

phoneCatalog.ApplyBulkDiscount(5);


ProductCatalog<Product> generalCatalog = new ProductCatalog<Product>();
generalCatalog.AddItem(new Electronics("Микроволновка", 8000, "Samsung", 12));
generalCatalog.AddItem(new Clothing("Куртка", 12000, "Nike", "L"));
generalCatalog.AddItem(new SmartWatch("Galaxy Watch 6", 25000, "Samsung", 12)); 

generalCatalog.DisplayAll();
generalCatalog.ApplyBulkDiscount(10);
```
